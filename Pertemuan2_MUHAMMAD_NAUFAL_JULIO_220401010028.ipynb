Python — List Operations

```python
# Membuat list
data = [10, 20, 999, True, 3.14]
# Akses & slicing
print(data[0]) # → 10
print(data[-1]) # → 3.14 (dari belakang)
print(data[1:3]) # → [20, 'Python']
# Manipulasi
data.append(99) # tambah di akhir
data.insert(1, 55) # sisipkan di posisi 1
data.remove(20) # hapus nilai pertama = 20
data.sort() # urutkan (jika tipe sama)
```

Python — NumPy Vektorisasi & Statistik

```python
# Operasi aritmetika vektorisasi
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])
print(a + b) # → [11 22 33 44 55]
print(a * b) # → [10 40 90 160 250]
print(a ** 3) # → [1 4 9 16 25]

# Universal functions (ufuncs)
np.sqrt(a) # akar kuadrat tiap elemen
np.exp(a) # eksponensial (e^x)
np.log(b) # logaritma natural

# Fungsi statistik
data = rng.normal(0, 1, 1000)
print(f'Mean: {np.mean(data):.4f}')
print(f'Std: {np.std(data):.4f}')
print(f'Min: {np.min(data):.4f}')
print(f'Max: {np.max(data):.4f}')
print(f'Q1: {np.percentile(data,25):.4f}')
print(f'Q3: {np.percentile(data,75):.4f}')
```

### Q1 Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

```python
missing = df.isnull().sum()
persentase = (missing / len(df)) * 100

hasil_missing = pd.DataFrame({
    'Jumlah Missing': missing,
    'Persentase (%)': persentase
})

print(hasil_missing)

print("\nJadi kolom yang memiliki missing values adalah Age, Cabin, dan Embarked")
print("Jumlah missing values pada kolom Age adalah 177, Cabin adalah 687, dan Embarked adalah 2")
print("Persentase missing values pada kolom Age adalah 19.875000%, Cabin adalah 77.100000%, dan Embarked adalah 0.222222%")
```

### Q3 Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.

```python
persen_selamat = df['Survived'].mean() * 100
print(f"Persentase penumpang selamat: {persen_selamat:.2f}%")
print("\n Jadi persentase dari penumpanh yang selamat adalah 38.38%")
```

### Q4 Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata-rata usianya?

```python
wanita_class1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

jumlah = len(wanita_class1)
rata_usia = wanita_class1['Age'].mean()

print("Jadi jumlah penumpanh wanita dari kelas 1 adalah :", jumlah)
print("dengan rata-rata usia:", rata_usia)
```

### Q5 Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

```python
survived = df.groupby('Pclass')['Survived'].mean()

print(survived)
print("\nTingkat keselamatan tertinggi terjadi pada kelas", survived.idxmax(), "dengan tingkat keselamatan sebesar ", survived.max())
```

### Q6 Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.

```python
print("Untuk analisis tambahan, saya menggunakan variabel kategori keluarga Besar, Kecil, dan Sendirian.")
print("Berikut data nya :")
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

def kategori_keluarga(x):
    if x == 1:
        return "Sendiri"
    elif x <= 4:
        return "Kecil"
    else:
        return "Besar"

df["Kategori Keluarga"] = df["FamilySize"].apply(kategori_keluarga)

hasil = df.groupby("Kategori Keluarga")["Survived"].mean() * 100

print(hasil)
print("Catatan : Keluarga kecil adalah keluarga kurang dari 4 orang, sedangkan Keluarga Besar adalah keluarga lebih dari 4 orang")
print("\nPenumpang yang bepergian dengan keluarga kecil cenderung memiliki tingkat keselamatan lebih tinggi")
print("dibandingkan penumpang yang bepergian sendiri atau dalam keluarga besar.")
print("Hal ini mungkin disebabkan karena anggota keluarga kecil dapat saling membantu saat evakuasi,")
print("sedangkan keluarga besar lebih sulit untuk tetap bersama dalam situasi darurat.")
```

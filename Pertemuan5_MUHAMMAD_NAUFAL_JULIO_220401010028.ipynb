```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('toyota.csv', sep=';')

df.head()
```

```python
df.shape
```

Dataset terdiri dari 100 baris dan 7 kolom yang merekam data penjualan mobil Toyota.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Grafik 1: Horizontal Bar Chart
sales_model = df.groupby('model')['total_sales'].sum().sort_values(ascending=True)

sales_model.plot(kind='barh', ax=axes[0, 0])

axes[0, 0].set_title('Grafik 1: Total Penjualan Berdasarkan Model Toyota')
axes[0, 0].set_xlabel('Total Sales (Juta Rupiah)')
axes[0, 0].set_ylabel('Model')


# Grafik 2: Histogram
axes[0, 1].hist(df['price'], bins=10)

axes[0, 1].set_title('Grafik 2: Distribusi Harga Mobil Toyota')
axes[0, 1].set_xlabel('Harga (Juta Rupiah)')
axes[0, 1].set_ylabel('Frekuensi')


# Grafik 3: Boxplot
df.boxplot(column='total_sales', by='sales_type', ax=axes[1, 0])

axes[1, 0].set_title('Grafik 3: Total Sales berdasarkan Jenis Pembayaran')
axes[1, 0].set_xlabel('Sales Type')
axes[1, 0].set_ylabel('Total Sales (Juta Rupiah)')


# Grafik 4: Scatter Plot
axes[1, 1].scatter(df['price'], df['qty'])

axes[1, 1].set_title('Grafik 4: Harga Mobil vs Jumlah Unit Terjual')
axes[1, 1].set_xlabel('Price (Juta Rupiah)')
axes[1, 1].set_ylabel('Qty')


# Menghilangkan judul otomatis dari boxplot
plt.suptitle('')

# Mengatur jarak antar grafik
plt.tight_layout()

# SIMPAN DULU
plt.savefig('dashboard_toyota.png', dpi=300, bbox_inches='tight')

# BARU TAMPILKAN
plt.show()
```

=============================================================

### Grafik 1. Total Penjualan Berdasarkan Model Toyota

### What?
Grafik menunjukkan bahwa **Corolla Cross Hybrid** memiliki total penjualan tertinggi, yaitu sebesar **13.640 juta rupiah**, sedangkan **Agya** memiliki total penjualan terendah, yaitu sebesar **3.240 juta rupiah**. Selain itu, model **Innova Reborn** dan **Veloz** juga menunjukkan performa penjualan yang cukup tinggi dengan total penjualan masing-masing **9.240 juta rupiah** dan **8.320 juta rupiah**.

### So what?
Perbedaan total penjualan antar model menunjukkan bahwa tidak semua model Toyota memberikan kontribusi yang sama terhadap pendapatan. Model dengan total penjualan tinggi dapat menjadi fokus utama dalam strategi pemasaran dan pengelolaan persediaan. Informasi ini penting agar dealer dapat mengalokasikan sumber daya secara lebih efektif.

### Now what?
Dealer dapat melakukan analisis lebih lanjut untuk mengetahui faktor-faktor yang menyebabkan tingginya penjualan Corolla Cross Hybrid, seperti tren pasar, fitur kendaraan, atau efektivitas promosi. Selain itu, perlu dievaluasi strategi pemasaran untuk model dengan penjualan rendah seperti Agya agar daya tariknya dapat meningkat.

=============================================================

### Grafik 2. Distribusi Harga Mobil Toyota

### What?
Histogram menunjukkan bahwa harga mobil Toyota dalam dataset berkisar antara **180 juta rupiah hingga 920 juta rupiah**. Nilai tengah (median) harga mobil adalah **320 juta rupiah**, sementara rata-rata harga mobil mencapai **383,8 juta rupiah**. Sebanyak 50% mobil memiliki harga di bawah **320 juta rupiah**, sedangkan 75% mobil memiliki harga di bawah **455 juta rupiah**.

### So what?
Distribusi harga ini memberikan gambaran mengenai segmen pasar yang dilayani oleh Toyota, mulai dari kelas entry-level hingga premium. Dealer dapat memahami bahwa sebagian besar produk berada pada rentang harga menengah, sehingga strategi penjualan perlu disesuaikan dengan daya beli konsumen pada segmen tersebut.

### Now what?
Analisis lanjutan dapat dilakukan untuk mengetahui hubungan antara harga mobil dengan total penjualan atau preferensi metode pembayaran. Selain itu, dealer dapat mengevaluasi apakah segmen harga tertentu memberikan kontribusi pendapatan yang lebih besar dibandingkan segmen lainnya.

=============================================================

### Grafik 3. Distribusi Total Sales Berdasarkan Jenis Pembayaran

### What?
Boxplot menunjukkan bahwa transaksi dengan metode **Credit** memiliki median total penjualan yang sedikit lebih tinggi, yaitu **750 juta rupiah**, dibandingkan transaksi **Cash** dengan median **640 juta rupiah**. Rata-rata total penjualan untuk transaksi Credit adalah **771,11 juta rupiah**, sedangkan transaksi Cash memiliki rata-rata **766 juta rupiah**. Nilai maksimum total penjualan pada kedua metode pembayaran sama-sama mencapai **2.760 juta rupiah**.

### So what?
Hasil ini menunjukkan bahwa metode pembayaran tidak memberikan perbedaan yang terlalu besar terhadap nilai transaksi penjualan. Namun, transaksi kredit cenderung memiliki nilai tengah yang lebih tinggi, yang dapat mengindikasikan bahwa konsumen lebih nyaman membeli kendaraan bernilai besar melalui skema cicilan.

### Now what?
Dealer dapat mengevaluasi program pembiayaan yang ditawarkan kepada konsumen untuk meningkatkan daya tarik pembelian secara kredit. Selain itu, analisis lebih lanjut dapat dilakukan untuk mengetahui apakah jenis kendaraan tertentu lebih banyak dibeli menggunakan metode pembayaran tertentu.

=============================================================

### Grafik 4. Hubungan Harga Mobil dan Jumlah Unit Terjual

### What?
Scatter plot menunjukkan bahwa hubungan antara harga mobil dan jumlah unit terjual tidak membentuk pola linear yang kuat. Hal ini didukung oleh nilai korelasi antara harga dan jumlah unit terjual yang hanya sebesar **0,06**, yang berarti hubungan keduanya sangat lemah. Mobil dengan harga tinggi maupun rendah dapat terjual dalam jumlah yang bervariasi.

### So what?
Temuan ini menunjukkan bahwa harga bukan satu-satunya faktor yang memengaruhi jumlah unit yang terjual. Faktor lain, seperti kebutuhan konsumen, reputasi model kendaraan, fitur yang ditawarkan, atau promosi penjualan, kemungkinan memiliki pengaruh yang lebih besar terhadap keputusan pembelian.

### Now what?
Penelitian lebih lanjut dapat dilakukan dengan memasukkan variabel lain, seperti model kendaraan, tahun produksi, atau metode pembayaran, untuk memahami faktor-faktor yang memengaruhi jumlah unit terjual secara lebih komprehensif. Dealer juga dapat menggunakan hasil ini untuk menyusun strategi pemasaran yang tidak hanya berfokus pada aspek harga.
